```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 14000, 18000, 22000, 25000],
    "Customers": [120, 145, 135, 170, 210, 240],
    "Profit": [3000, 4000, 3500, 5000, 6500, 7500]
}

df = pd.DataFrame(data)

df
```

```python
plt.figure(figsize=(8,5))

plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Customers"], df["Sales"])

plt.title("Customers vs Sales")
plt.xlabel("Number of Customers")
plt.ylabel("Sales (₹)")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(df["Month"], df["Profit"])

plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit (₹)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(df["Month"], df["Sales"], marker="o", label="Sales")
plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")

plt.title("Monthly Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount (₹)")
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(7,5))

correlation = df[["Sales", "Customers", "Profit"]].corr()

sns.heatmap(correlation, annot=True, cmap="Blues")

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("KEY INSIGHTS")
print("1. Sales increased consistently from January to June.")
print("2. The number of customers increased along with sales.")
print("3. Profit also increased during the six-month period.")
print("4. Sales, customers, and profit show a strong positive relationship.")
```

```python
print("CONCLUSION")
print("Data visualization helped us understand sales, customer growth,")
print("and profit trends clearly. Charts and heatmaps made the")
print("patterns and relationships in the data easy to identify.")
```
